## <a name="coding"></a> 2. Propositional Logic Coding Problems (28 points)

In this problem, you will implement the DPLL algorithm we discussed in Lecture 6. Since the DPLL algorithm expects sentences in CNF form, we provide a few helper functions that you will make use of, which you should have a look at:

````python
def lit_to_var_val(literal: int) -> tuple[int, bool]:
    """Converts a literal into (variable, value).

    Args:
      literal: A nonzero int.

    Returns:
      variable: A positive int representing a variable.
      value: True or False, i.e., positive or negative.
    """
    return abs(literal), literal > 0


def is_cnf_formula(formula: list) -> bool:
    """Checks whether the input is a valid CNF formula.

    A formula is in valid CNF form if it is a list of lists of nonzero
    integers with sign indicating whether the variable is negated.

    You will not need to use this utility in your implementation,
    but it may be useful to read to understand the CNF representation.
    """
    if not isinstance(formula, list):
        return False
    if len(formula) == 0:
        return True
    clause = formula[0]
    if not isinstance(clause, list):
        return False
    for literal in clause:
        if not isinstance(literal, int):
            return False
        if literal == 0:
            return False
    if len(formula) == 1:
        return True
    return is_cnf_formula(formula[1:])


def get_variables_in_cnf_formula(cnf_formula: list) -> list[int]:
    """Get a list of all variables in a CNF formula.

    Args:
      cnf_formula: A list of lists of nonzero ints.

    Returns:
      variables: A list of all variables that appear in
        the formula.
    """
    variables = set()
    for clause in cnf_formula:
        variables.update({lit_to_var_val(literal)[0] for literal in clause})
    variables = sorted(variables)
    return variables
````

### <a name="dpll"></a> 2.2 DPLL (23 points)

You will now implement DPLL using the helper functions below, which you should have a look at:

````python
def clause_is_determined(clause: list[int], partial_assignment: dict[int, bool]) -> bool:
    """Checks whether all variables in the clause have an assignment.

    Args:
      clause: A list of nonzero ints.
      partial_assignment: A dict of variables (ints) to values (bools).

    Returns:
      is_determined: True if all variables in the clause appear in
        partial_assignment.
    """
    for literal in clause:
        if not (literal in partial_assignment or
                -literal in partial_assignment):
            return False
    return True


def literal_is_satisfied(literal: int, partial_assignment: dict[int, bool]) -> bool:
    """Checks whether the literal is satisfied by the assignment.

    Args:
      literal: A nonzero int.
      partial_assignment: A dict of variables (ints) to values (bools).

    Returns:
      is_satisfied: True if the literal's variable appears in the
        partial_assignment, with a sign matching the literal.
    """
    variable, val = lit_to_var_val(literal)
    return variable in partial_assignment and partial_assignment[variable] == val


def clause_is_satisfied(clause: list[int], partial_assignment: dict[int, bool]) -> bool:
    """Checks whether the clause is satisfied by the assignment.

    Args:
      clause: A list of nonzero ints.
      partial_assignment: A dict of variables (ints) to values (bools).

    Returns:
      is_satisfied: True if some literal in the clause is satisfied.
    """
    for literal in clause:
        if literal_is_satisfied(literal, partial_assignment):
            return True
    return False


def find_pure_variable(
    cnf_formula: list,
    variables: list[int],
    partial_assignment: dict[int, bool] | None) -> tuple[int | None, bool | None]:
    """Helper for DPLL.

    A variable is pure if it has the same sign in all unsatisfied clauses
    and if it is not already assigned.

    If a pure variable exists, this function returns the variable and value
    corresponding to the literal. (If multiple exist, return an arbitrary one.)

    If no pure variables exist, return (None, None).

    Args:
      cnf_formula: A list of lists of nonzero integers representing a CNF formula,
        with sign indicating whether the variable is negated.
      variables: A list of positive integers.
      partial_assignment: A dict mapping positive integers to bools, or None for
        an empty assignment.

    Returns:
      variable : A positive integer or None.
      value: A bool or None.
    """
    candidate_to_possible_values  = {v : {True, False} for v in variables \
                                    if v not in partial_assignment}
    for clause in cnf_formula:
        if clause_is_satisfied(clause, partial_assignment):
            continue
        for literal in clause:
            variable, value = lit_to_var_val(literal)
            if variable in candidate_to_possible_values:
                candidate_to_possible_values[variable].discard(not value)
    for candidate, possible_values in candidate_to_possible_values.items():
        if possible_values:
            value = next(iter(possible_values))
            return candidate, value
    return None, None


def find_unit_clause(
        cnf_formula: list,
        partial_assignment: dict[int, bool] | None) -> tuple[int | None, bool | None]:

    """Helper for DPLL.

    A clause is a unit clause if all literals but one are already assigned to
    False. If a unit clause exists, this function returns the variable and value
    corresponding to the literal. (If multiple exist, return an arbitrary one.)

    Args:
      cnf_formula: A list of lists of nonzero integers representing a CNF formula,
        with sign indicating whether the variable is negated.
      variables: A list of positive integers.
      partial_assignment: A dict mapping positive integers to bools, or None for
        an empty assignment.

    Returns:
      variable : A positive integer or None.
      value: A bool or None.
    """
    for clause in cnf_formula:
        unassigned_literal = None
        is_unit_clause = True
        for literal in clause:
            # If the literal is true in the assignment, this is not a unit clause
            if literal_is_satisfied(literal, partial_assignment):
                is_unit_clause = False
                break
            # If the literal is false in the assignment, this could be a unit clause
            elif literal in partial_assignment:
                continue
            # If there is already an unassigned literal, this is not a unit clause
            elif not (unassigned_literal is None):
                is_unit_clause = False
                break
            else:
                unassigned_literal = literal
        if is_unit_clause and not (unassigned_literal is None):
            return lit_to_var_val(unassigned_literal)
    return None, None
````

Complete an implementation of DPLL, which takes a `cnf_formula` and finds a satisfying assignment of variables.

For reference, our solution is **62** line(s) of code.

````python
def run_dpll(cnf_formula: list,
    unassigned_variables: list[int],
    partial_assignment: dict[int, bool] | None) -> dict[int, bool] | None:
    # if every clause in clauses is true in partmodel then return true
    all_satisfied = True
    for clause in cnf_formula:
        if not clause_is_satisfied(clause, partial_assignment):
            all_satisfied = False
            if clause_is_determined(clause, partial_assignment):
                # we have come across an impossible to satisfy clause
                return None

    if all_satisfied:
        return partial_assignment

    # Pure variables
    pure_variable, value = find_pure_variable(cnf_formula, unassigned_variables, partial_assignment)
    if pure_variable is not None and pure_variable not in partial_assignment:
        new_assignments = {**partial_assignment, pure_variable: value}
        assignment = run_dpll(cnf_formula, [var for var in unassigned_variables if var != pure_variable], new_assignments)
        if assignment is not None:
            return assignment

    # Unit clauses where setting a variable to True must happen for the formula to work
    unit_variable, value = find_unit_clause(cnf_formula, partial_assignment)
    if unit_variable is not None and unit_variable not in partial_assignment:
        new_assignments = {**partial_assignment, unit_variable: value}
        assignment = run_dpll(cnf_formula, [var for var in unassigned_variables if var != unit_variable], new_assignments)
        if assignment is not None:
            return assignment

    # Now we just need to search the child nodes
    child_variable = unassigned_variables[0]

    # try setting variable to True
    new_assignments = {**partial_assignment, child_variable: True}
    true_attempt = run_dpll(cnf_formula, unassigned_variables[1:], new_assignments)
    
    if true_attempt is not None:
        return true_attempt

    # try setting variable to False
    new_assignments = {**partial_assignment, child_variable: False}
    false_attempt = run_dpll(cnf_formula, unassigned_variables[1:], new_assignments)

    return false_attempt

def run_inference_dpll(cnf_formula: list) -> tuple[bool, dict[int, bool] | None]:
    """Find a satisfying assignment for a propositional CNF formula with DPLL.

    Args:
      cnf_formula: A list of lists of nonzero integers representing a CNF formula,
        with sign indicating whether the variable is negated.

    Returns:
      satisfiable: A bool indicating whether some satisfying assignment exists.
      assignment: A dict mapping positive integers to bools, or None if no
        satisfying assignment exists.

    Examples:
      >> run_inference_dpll([[1, -2], [-1, -2]])
      >> (True, {1: True, 2: False}))

      >> run_inference_dpll([[1], [-1]])
      >> (False, None)
    """
    if not is_cnf_formula(cnf_formula):
        raise ValueError("Invalid CNF Formula given")
    variables = get_variables_in_cnf_formula(cnf_formula)
    assignments = run_dpll(cnf_formula, variables, {})
    if assignments is None:
        return (False, None)

    # Give some default True value to variables that don't matter
    for var in variables:
        if var not in assignments:
            assignments[var] = True
            
    return (True, assignments)
    
````

Use sympy to determine whether the following formula is satisfiable:
$(\neg x_1 \land x_2) \Rightarrow ((x_2 \lor x_3) \land (x_1 \lor \neg x_3))$. You will need to create the formula and then use sympy to check if it is satisfiable. 

Note that you should return two things: a **bool** and the sympy formula you create.

For reference, our solution is **3** line(s) of code.

````python
def formula1_is_satisfiable() -> tuple[bool, sympy.Basic]:
    """Determines whether the above formula is satisfiable.

    Returns:
      is_satisfiable: A bool indicating whether the formula is satisfiable.
      formula: the sympy formula that you create.
    """
    x1, x2, x3 = sympy.symbols('x1 x2 x3')
    formula = sympy.Implies(sympy.And(sympy.Not(x1), x2), sympy.And(sympy.Or(x2, x3), sympy.Or(x1, sympy.Not(x3))))
    return bool(sympy.logic.inference.satisfiable(formula)), formula
````

### <a name="searchrescue"></a> 3.3 Search and Rescue Inference (19 points)

Now, we will look at the actual inference problem that we are interested in. We will consider several problems with varying grid sizes and different sets of observations. For example, consider the grid below:
```
# Fire, Unknown, Clear, Smoke
GRID0 = np.array([
  ["F", "U", "C"],
  ["S", "C", "U"],
  ["U", "U", "C"]
], dtype=object)
```
This grid has 9 locations and 5 observations: there is fire in the top left, smoke below it, and the center, top right, and bottom right locations are all known to be clear of smoke or fire.

We will assume the following axioms:
1. Each location has exactly one of {smoke, fire, clear}.
2. There is smoke at a location if and only if there is a fire in at least one of the adjacent (above, below, left, right) locations. Diagonals are not adjacent!

Note that as a corollary, there cannot be two fires in adjacent locations. Take a moment to run your human inference engine: which unknown values in the grid above can be determined?

Now, write a program that takes a grid as input and infers unknown values. Your program should output a new grid with all determinable unknown values replaced with the inferred value. If an unknown value cannot be determined, it should be left unknown.

**Your program should use sympy.**


For reference, our solution is **57** line(s) of code.

````python
def infer_unknown_values(grid: list) -> list:
    """Fill in any unknown values in the grid that can be inferred.

    Args:
      grid: A list of lists of "F", "U", "S", or "C".

    Returns:
      inferred_grid: A copy of grid with some unknown values replaced.

    Example:
      >> grid = [
      >>   ["F", "U", "C"],
      >>   ["S", "C", "U"],
      >>   ["U", "U", "C"]
      >> ]
      >> infer_unknown_values(grid)
      >> [["F" "S" "C"]
      >>  ["S" "C" "C"]
      >>  ["U" "U" "C"]]
    """
    F = {}
    S = {}
    C = {}

    # setup symbolic grid
    for r in range(len(grid)):
        for c in range(len(grid[0])):
            F[(r, c)] = sympy.symbols(f"F_{r}_{c}")
            S[(r, c)] = sympy.symbols(f"S_{r}_{c}")
            C[(r, c)] = sympy.symbols(f"C_{r}_{c}")

    known_clauses = []

    
    
    for row in range(len(grid)):
        for col in range(len(grid[0])):
            value = grid[row][col]
            fire = F[(row, col)]
            smoke = S[(row, col)]
            clear = C[(row, col)]

            # add known truths from direct observations
            if value == "F":
                known_clauses.append(fire)
            if value == "S":
                known_clauses.append(smoke)
            if value == "C":
                known_clauses.append(clear)

            # at least one must be true
            known_clauses.append(fire | smoke | clear)

            # at most one can be true
            known_clauses.append(~(fire & smoke))
            known_clauses.append(~(fire & clear))
            known_clauses.append(~(smoke & clear))

            # find adjacent fires
            adjacent_fires = []
            if row > 0:
                adjacent_fires.append(F[(row - 1, col)])
            if row < len(grid) - 1:
                adjacent_fires.append(F[(row + 1, col)])
            if col > 0:
                adjacent_fires.append(F[(row, col - 1)])
            if col < len(grid[0]) - 1:
                adjacent_fires.append(F[(row, col + 1)])

            # smoke here IFF one of the adjacent tiles are a fire
            if adjacent_fires:
                known_clauses.append(sympy.Equivalent(smoke, sympy.Or(*adjacent_fires)))
            else:
                known_clauses.append(~smoke)

            
    knowledge_base = sympy.And(*known_clauses)

    # input validation, check the input is valid against the constraints before making changes
    if not sympy.logic.inference.satisfiable(knowledge_base):
        raise ValueError("Provided grid is invalid!")

    # make a copy of the grid that we can return
    inferred_grid = [row[:] for row in grid]
    
    # go through each tile and determine whether we can assign a value
    for row in range(len(grid)):
        for col in range(len(grid[0])):
            value = grid[row][col]
            fire = F[(row, col)]
            smoke = S[(row, col)]
            clear = C[(row, col)]

            if value == "U":
                # See if any of these values work
                if not sympy.logic.inference.satisfiable(knowledge_base & ~fire):
                    inferred_grid[row][col] = "F"
                    knowledge_base = knowledge_base & fire
                elif not sympy.logic.inference.satisfiable(knowledge_base & ~smoke):
                    inferred_grid[row][col] = "S"
                    knowledge_base = knowledge_base & smoke
                elif not sympy.logic.inference.satisfiable(knowledge_base & ~clear):
                    inferred_grid[row][col] = "C"
                    knowledge_base = knowledge_base & clear

            
    return inferred_grid
````

## <a name="pacman_kb"></a> 4. Build a Pac-Man Knowledge Base [AI-enabled] (8 points)

Use an LLM (ChatGPT, Claude, etc.) to develop a propositional knowledge base for partially observable Pac-Man. Implement `build_pacman_kb(grid, actions, percepts, start=None)` below using SymPy, as in Section 3. 

Your function should return **one SymPy formula whose satisfying assignments are exactly the worlds consistent with these rules**:

1. The map: `grid` is a nonempty rectangular list of lists containing only `"#"` (wall) and `"."` (open). The entire map is known and does not change over time; outside the grid is always blocked. Pac-Man may still be uncertain about its position: different open cells can have the same neighboring-wall observations.
2. Pac-Man's position: Pac-Man occupies exactly one open cell at every time. Coordinates are `(row, col)`, with `(0, 0)` at the top left. `start` gives its initial position; `None` means the initial position is unknown.
3. Pac-Man's Movement: `actions[t]` is the attempted move from time `t` to `t + 1`: `"North"`, `"South"`, `"East"`, or `"West"`. North/South decrease/increase the row; East/West increase/decrease the column. Pac-Man moves to the neighboring cell if it is open, and stays in place if it is blocked.
4. Pac-Man's observations: Pac-Man makes an observation **initially and after every action**, including when a move is blocked. `percepts` is a **list with one tuple of four booleans per observation**: `(north_blocked, south_blocked, east_blocked, west_blocked)`. The direction order is **North, South, East, West**. A `True` means that immediate neighbor is blocked (a wall or outside the grid); a `False` means it is open. Readings are exact. For `T = len(actions)`, `percepts` has `T + 1` tuples: `percepts[0]` is the initial observation, and `percepts[t + 1]` is the observation after `actions[t]`. Thus `percepts[T]` describes the final position.

For example, two actions give three observations:

```python
actions = ["East", "South"]
percepts = [
    (True, False, False, True),   # t=0: initial observation, before East
    (True, False, False, False),  # t=1: after East, before South
    (False, True, True, False),   # t=2: after South (the final position)
]
```

Here `percepts[0][0]` is `True`, so North is blocked initially; `percepts[0][2]` is `False`, so East is open initially. The outer list indexes time, and each inner tuple indexes direction.

Use the provided helpers to create propositions:
 - `pacman_wall(row, col)`: cell `(row, col)` is a wall, with symbol `W_row_col`. 
 - `pacman_at(row, col, t)`: Pac-Man is at `(row, col)` at time `t`, with symbol `P_row_col_t`. 

A complete world assigns a truth value to every in-grid wall proposition and every position proposition at times `0, ..., T`, including positions in wall cells. Use only these propositions; actions and observations are input data. You may assume correctly formatted inputs and an in-bounds `start`. Inconsistent evidence must produce an unsatisfiable formula. 

*Suggestion*: Use the autograder as feedback for your LLM. The autograder will raise failing world instances, which you can use as re-prompting feedback to improve the LLM's generated KB.


````python
# Provided vocabulary. Do not change these helpers or the direction convention.
PACMAN_DIRECTIONS = {
    "North": (-1, 0), "South": (1, 0),
    "East": (0, 1), "West": (0, -1),
}


def pacman_wall(row, col):
    return sympy.Symbol(f"W_{row}_{col}")


def pacman_at(row, col, t):
    return sympy.Symbol(f"P_{row}_{col}_{t}")
````

<div class="alert alert-info">
Write your implementation in the function below. 
</div>

````python
def build_pacman_kb(grid, actions, percepts, start=None):
    """Return a SymPy KB for position histories consistent with the known map and evidence."""
    rows = len(grid)
    cols = len(grid[0])
    T = len(actions)

    known_clauses = []

    # 1. Map constraints: grid walls are fully known
    for row in range(rows):
        for col in range(cols):
            wall = pacman_wall(row, col)
            if grid[row][col] == "#":
                known_clauses.append(wall)
            else:
                known_clauses.append(~wall)

    # 2. Position constraints at every time step t
    for t in range(T + 1):
        open_positions = []
        for row in range(rows):
            for col in range(cols):
                pos = pacman_at(row, col, t)
                if grid[row][col] == "#":
                    # Pac-Man can never be in a wall
                    known_clauses.append(~pos)
                else:
                    open_positions.append(pos)

        # at least one open position must be occupied
        known_clauses.append(sympy.Or(*open_positions))

        # at most one position can be occupied
        for i in range(len(open_positions)):
            for j in range(i + 1, len(open_positions)):
                known_clauses.append(~(open_positions[i] & open_positions[j]))

    # 3. Known initial position if provided
    if start is not None:
        start_row, start_col = start
        known_clauses.append(pacman_at(start_row, start_col, 0))

    # 4. Sensor observations (Percepts)
    for t in range(T + 1):
        north_blocked, south_blocked, east_blocked, west_blocked = percepts[t]

        for row in range(rows):
            for col in range(cols):
                if grid[row][col] == ".":
                    pos = pacman_at(row, col, t)

                    # check if the neighbor cells are blocked by walls or outside grid
                    is_n_blocked = (row == 0) or (grid[row - 1][col] == "#")
                    is_s_blocked = (row == rows - 1) or (grid[row + 1][col] == "#")
                    is_e_blocked = (col == cols - 1) or (grid[row][col + 1] == "#")
                    is_w_blocked = (col == 0) or (grid[row][col - 1] == "#")

                    matches_percept = (
                        is_n_blocked == north_blocked
                        and is_s_blocked == south_blocked
                        and is_e_blocked == east_blocked
                        and is_w_blocked == west_blocked
                    )

                    # if readings here do not match the observation, Pac-Man cannot be here
                    if not matches_percept:
                        known_clauses.append(~pos)

    # 5. Transition model: moving between time steps
    for t in range(T):
        action = actions[t]
        dr, dc = PACMAN_DIRECTIONS[action]

        for row in range(rows):
            for col in range(cols):
                if grid[row][col] == ".":
                    curr_pos = pacman_at(row, col, t)

                    next_row = row + dr
                    next_col = col + dc

                    # if the target cell is inside the grid and open, Pac-Man moves there
                    if 0 <= next_row < rows and 0 <= next_col < cols and grid[next_row][next_col] == ".":
                        dest_pos = pacman_at(next_row, next_col, t + 1)
                    else:
                        # otherwise the move is blocked and Pac-Man stays put
                        dest_pos = pacman_at(row, col, t + 1)

                    # Pac-Man being here at time t implies being at dest at t + 1
                    known_clauses.append(curr_pos >> dest_pos)

    # combine into one final knowledge base
    knowledge_base = sympy.And(*known_clauses)

    return knowledge_base
````
